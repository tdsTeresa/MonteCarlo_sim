 
### Simulación Monte Carlo:<br>
#### Cálculo de costos de producción en un negocio abarrotero para diciembre

```python
# Importación de librerías Numpy y random para generar las distribuciones
import numpy as np
import random
```

```python
# Definición de variables de presupuesto y el costo de referencia de este mes en pesos
costo_renta= 0
costo_sueldos= 0
costo_servicios= 0
costo_mercancia= 0
costo_aguinaldos= 0 
costo_regalo_clientes= 0
costo_referencia= 70000

```

```python
# Definición del número de iteraciones 
max_simulations= 5000
```

```python
# Función con distribuciones normales de costos.
# Las variables son: renta, sueldo de dos empleados, pago de servicios (luz, internet, agua), compra de mercancía, aguinaldo y regalos de fin de año.
def costos_simulation():
    costo_renta= np.random.normal(6000, 100)
    costo_sueldos= np.random.normal(17000, 1000)
    costo_servicios= np.random.normal(5000, 500)
    costo_mercancia= np.random.normal(25000, 1000)
    costo_aguinaldos= np.random.normal(8500, 200)
    costo_regalo_clientes= np.random.normal(10000, 300)
    return costo_renta, costo_sueldos, costo_servicios, costo_mercancia, costo_aguinaldos, costo_regalo_clientes
```

```python
# Variables para almacenar la suma de los items y el número de iteraciones
balance= [costo_referencia]
num_simulations= [0]
```

```python
type(balance)
```

```python
while num_simulations[-1] < max_simulations:                                # Ciclo que se ejecuta definido por el número máximo de iteraciones
    iter= costos_simulation()                                               # variable que almacena la función
    costo_total= iter[0] + iter[1] + iter[2] + iter[3] + iter[4] + iter[5]  # Suma de los posibles costos de cada item de la función
    balance.append(costo_total)                                             # Se concatena la suma anterior en la lista balance, que incluye el presupuesto del mes
    num_simulations.append(num_simulations[-1] + 1)                         # Se incrementa en uno cada iteración y se almacena en la lista
```

```python
balance
```

```python
# Longitud de la lista balance
balance.__len__() 
```

```python
import matplotlib.pyplot as plt     # gráfico de línea de las 5,000 simulaciones
plt.plot(num_simulations, balance)
plt.title('Cálculo de costos de producción en diciembre')
plt.xlabel('Iteraciones')
plt.ylabel('Costo en pesos')
plt.show()
```

```python
# Definición de estilo de gráfico
plt.style.use('bmh')
```

```python
# Obtenemos una lista con los costos menores a $ 70,000
costos = [x for x in balance if x < costo_referencia]
costos
```

```python
# Cálculo de la probabilidad de que el costo total sea menor al presupuesto
probabilidad= (costos.__len__() / balance.__len__()) *100
```

```python
print(f' La probabilidad de que el costo total sea menor al costo de referencia es {probabilidad:.2f} %')
```

```python
# Histograma del presupuesto
plt.hist(balance, density= True, bins= 80)
plt.title('Histograma de simulación de presupuesto')
plt.axvline(np.mean(balance), color= 'k', linestyle= 'dashed', linewidth= 1) # Línea punteada que marca la media de presupuestos obtenidos en la simulación
plt.axvline(costo_referencia, color= 'r', linestyle= 'dashed', linewidth= 1)    # Línea roja que marca el presupuesto inicial que se tenía planeado
plt.xlabel('Presupuesto')
plt.ylabel('Frecuencia')
plt.show()
```

```python
# Valores mínimo y máximo del gráfico
print('Mínimo: ', min(balance), '   Máximo: ', max(balance))
```

```python
# Percentiles de la distribución de resultados de la simulación
for i in range(0, 100, 5):
    print('Percentil ', i, '% = ', np.percentile(balance, i))
```

```python
# El presupuesto inicial ($ 70,000) está en la posición del 16 %
for i in range(15, 20, 1):
   print('Percentil ', i, '% = ', np.percentile(balance, i)) 
```

#### Resultados:
- El presupuesto dado por las simulaciones va de 65, 180 a  77,786 pesos, lo que significa que en el 16 % el costo total es menor o igual al presupuesto inicial de $ 70,000.

### Análisis de sensibilidad con dos distribuciones

```python
# Las variables a usar serán las mismas, excepto por balance_2, donde se almacenarán los resultados
costo_referencia= 70000
balance_2= [costo_referencia]
num_simulations= [0]
max_simulations= 5000
```

```python
# Se realizan ajustes en los costos de producción ahora con diferentes tipos de distribución
def costos_simulation2():
    costo_renta= np.random.uniform(6000, 6200)                      # Existe la misma probabilidad de pagar $ 6,000 a $6,200 de renta de local
    costo_sueldos= np.random.normal(17000, 800)                     # Se toman en cuenta los posibles bonos (desviación estándar)
    costo_servicios= np.random.triangular(5000, 5200, 6000)         # Se toman en cuenta los riesgos de costos extremos en la variación de servicios
    costo_mercancia= np.random.triangular(20000, 25000, 27000)      # Se toman en cuenta los riesgos de costos extremos en el pago de mercancía
    costo_aguinaldos= np.random.normal(8500, 100)
    costo_regalo_clientes= np.random.triangular(9000, 9500, 10000)   # El presupuesto de regalo de fin de año ha cambiado 
    return costo_renta, costo_sueldos, costo_servicios, costo_mercancia, costo_aguinaldos, costo_regalo_clientes
```

```python
costos_simulation2()
```

```python
while num_simulations[-1] < max_simulations:                                # Ciclo que se ejecuta definido por el número máximo de iteraciones
    iter= costos_simulation2()                                               # variable que almacena la función
    costo_total_2= iter[0] + iter[1] + iter[2] + iter[3] + iter[4] + iter[5]  # Suma de los posibles costos de cada item de la función
    balance_2.append(costo_total_2)                                             # Se concatena la suma anterior en la lista balance_2, que incluye el presupuesto del mes
    num_simulations.append(num_simulations[-1] + 1)  
```

```python
    # gráfico de línea de las 5,000 simulaciones
plt.plot(num_simulations, balance_2)
plt.title('Segundo cálculo de costos en diciembre')
plt.xlabel('Iteraciones')
plt.ylabel('Costo en pesos')
plt.show()
```

```python
# Valores mínimo y máximo del nuevo gráfico
print('Mínimo: ', min(balance_2), '   Máximo: ', max(balance_2))
```
